## Part 1 – Feature Engineering
### Exercise 1: Create new features from existing ones
**Dataset:** Titanic dataset (`seaborn.load_dataset('titanic')`)

```python
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Load Titanic dataset
titanic = sns.load_dataset('titanic')
#titanic.head()

# 1. Create FamilySize = sibsp + parch + 1
titanic['FamilySize'] = titanic['sibsp'] + titanic['parch'] + 1

# 2. Create IsAlone = 1 if FamilySize == 1 else 0
titanic['IsAlone'] = np.where(titanic['FamilySize'] == 1, 1, 0)

# 3. Convert 'sex' column into numerical format (0=female, 1=male)
titanic['sex'] = np.where(titanic['sex'] == 'male', 1, 0)

# 4. Extract the Title from names (if name column exists)
#titanic['name'] = titanic['who']
if 'name' in titanic.columns:
    titanic['Title'] = titanic['name'].str.extract(r' ([A-Za-z]+)\.', expand=False)

#titanic[['sibsp', 'parch', 'FamilySize']].head()

titanic.head()
```
