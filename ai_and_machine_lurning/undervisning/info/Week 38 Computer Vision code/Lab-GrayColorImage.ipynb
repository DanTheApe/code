# 2. Reading and displaying images

Reading images in OpenCV is very straightforward. We use the imread() function which allows us to read different types of images (JPG, PNG, etc…). You can read them as grayscale or color images. You can also read images that contain an Alpha channel.

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import matplotlib
# Tells notebook to render figures in-page.
%matplotlib inline
from IPython.display import Image
plt.rcParams['image.cmap'] = 'gray'

Image(filename='messi5.jpg')
```

## 3. Display Image Data as an "Image" using Matplotlib

```python
img = cv2.imread('haaland.jpg')
assert img is not None, "file could not be read, check with os.path.exists()"

plt.imshow(img)
print("Image size is ", img.shape)
```

### What happened?
1) The color representation is unexpected.
2) The displayed image is a mathematical representation of the image (essentially a "plot" of the image data).

Matplotlib uses different color maps to interpret pixel intensities. To properly render this image as a grayscale image we need to specify the proper colormap.

Matplotlib expects the image to be in RGB format whereas OpenCV stores images in BGR format. Thus, we need to reverse the channel order of the image in order to properly render the color of the image.

### Swap the Red and Blue channels
Let's create a copy of the image and call it img2. Then we use img2= img[:, :, ::-1] to swap the channels

```python
img2 = cv2.imread('haaland.jpg')
img2= img2[:, :, ::-1]
plt.imshow(img2)
```

## 4. Change the size of the picture

```python
plt.figure(figsize = (10, 10))
plt.imshow(img2)
```

## 5. Grayscale image

```python
img_gray = cv2.imread('haaland.jpg',cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray)
```

## 6. Splitting and merging color channels

```python
# Split the image into the B,G,R components.
img_bgr = cv2.imread('Emerald_Lakes_New_Zealand.jpg', cv2.IMREAD_COLOR)
b, g, r = cv2.split(img_bgr)

# Show the channels.
plt.figure(figsize = [20, 10])
plt.subplot(141); plt.imshow(r); plt.title('Red Channel')
plt.subplot(142); plt.imshow(g); plt.title('Green Channel')
plt.subplot(143); plt.imshow(b); plt.title('Blue Channel')

# Merge the individual channels into a BGR image.
imgMerged = cv2.merge((r, g, b))

# Display the merged output.
plt.subplot(144)
plt.imshow(imgMerged)
plt.title('Merged Output');
```
