# Haar Cascades for Object Detection

Haar Cascade classifiers are a machine learning-based method for object detection.

## 1.  Importing required Libraries and Haar Cascade Classifier


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier('haarcascade_eye.xml')
```

## 2. Creating a function to detect faces and eyes

Create a function **adjusted_detect_face()** to detect faces in an image using the face cascade classifier and **detect_eyes()** to detect eyes using the eye cascade classifier. These functions will draw rectangles around the detected faces and eyes.

Use the function **cv2.rectangle()**, this function draws a rectangle on the image at the coordinates (x, y) with a width w and height h. The color is white (255, 255, 255) and the rectangle has a thickness of 10 pixels.


```python
# Creating Function to Detect Faces

def adjusted_detect_face(img):
    face_img = img.copy()
    face_rect = face_cascade.detectMultiScale(face_img, scaleFactor=1.2, minNeighbors=5)

    for (x, y, w, h) in face_rect:
        cv2.rectangle(face_img, (x, y), (x + w, y + h), (255, 255, 255), 10)

    return face_img

# Creating Function to Detect Eyes
def detect_eyes(img):
    eye_img = img.copy()
    eye_rect = eye_cascade.detectMultiScale(eye_img, scaleFactor=1.2, minNeighbors=5)

    for (x, y, w, h) in eye_rect:
        cv2.rectangle(eye_img, (x, y), (x + w, y + h), (255, 255, 255), 10)

    return eye_img
```

## 3. Load the image

Load the image and create three copies of the images

```python
img = cv2.imread('andrew.jpg')
img_copy1 = img.copy()
img_copy2 = img.copy()
img_copy3 = img.copy()

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

## 4. Detecting Faces and Eyes

This code will output three images: **Face Detection, Eyes Detection and Face and Eyes Detection**. These images will also be saved as face.jpg, eyes.jpg and face+eyes.jpg respectively.




```python
# Face Detection
face = adjusted_detect_face(img_copy1)
plt.imshow(cv2.cvtColor(face, cv2.COLOR_BGR2RGB))
plt.show()
cv2.imwrite('face.jpg', face)

# Eyes Detection:
eyes = detect_eyes(img_copy2)
plt.imshow(cv2.cvtColor(eyes, cv2.COLOR_BGR2RGB))
plt.show()
cv2.imwrite('eyes.jpg', eyes)

# Face and Eyes Detection:
face3 = adjusted_detect_face(img_copy3)
eyes_face = detect_eyes(face3)
plt.imshow(cv2.cvtColor(eyes_face, cv2.COLOR_BGR2RGB))
plt.show()
cv2.imwrite('face+eyes.jpg', eyes_face)
```
